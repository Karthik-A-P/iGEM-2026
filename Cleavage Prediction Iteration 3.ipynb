```python
# Basic
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Train-test split and preprocessing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Models
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier

# XGBoost
from xgboost import XGBClassifier

# Evaluation
from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    roc_curve
)

print("All libraries imported successfully.")
```

```python
# ============================================
# PREPROCESSING + SPLIT
# ============================================

# ---- Target: explicitly map classes ----
data["Enzyme"] = data["Enzyme"].replace({
    "Aspartic": 0,
    "Serine": 1
})

# Make sure target is numeric
y = pd.to_numeric(data["Enzyme"], errors="coerce")

# Check for missing labels
if y.isna().any():
    print("Missing/invalid enzyme labels:")
    print(data.loc[y.isna(), "Enzyme"].value_counts(dropna=False))
    raise ValueError(
        "Enzyme contains missing/invalid labels. "
        "Reload the original data before preprocessing."
    )

y = y.astype(int)

# ---- Features ----
X = data.drop(columns=["Substrate", "Enzyme"])

# Keep only numerical features
non_numeric = X.select_dtypes(exclude=["number"]).columns

if len(non_numeric) > 0:
    print("Dropping non-numeric columns:")
    print(non_numeric.tolist())
    X = X.drop(columns=non_numeric)

# ---- Train/test split ----
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.40,
    random_state=42,
    stratify=y
)

# ---- Scaling ----
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# ============================================
# RESULTS
# ============================================

print("\n========== DATASET ==========")
print("Total samples    :", len(X))
print("Training samples :", len(X_train))
print("Testing samples  :", len(X_test))
print("Features         :", X.shape[1])

print("\n========== CLASS DISTRIBUTION ==========")
print("Aspartic (0):", (y == 0).sum())
print("Serine (1)  :", (y == 1).sum())

print("\n========== SPLIT ==========")
print("Training:")
print(y_train.value_counts().sort_index())

print("\nTesting:")
print(y_test.value_counts().sort_index())

print("\nScaling completed.")
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

# Impute missing values using training-set median
imputer = SimpleImputer(strategy="median")

X_train_imputed = imputer.fit_transform(X_train)
X_test_imputed = imputer.transform(X_test)

# Scale
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train_imputed)
X_test_scaled = scaler.transform(X_test_imputed)

print("NaNs in training:", np.isnan(X_train_scaled).sum())
print("NaNs in testing:", np.isnan(X_test_scaled).sum())
```

```python
print("X_train_scaled NaNs:", np.isnan(X_train_scaled).sum())
print("X_test_scaled NaNs:", np.isnan(X_test_scaled).sum())
print("X_train_scaled inf:", np.isinf(X_train_scaled).sum())
print("X_test_scaled inf:", np.isinf(X_test_scaled).sum())
```

```python
print("Shape:", X_train_scaled.shape)
print("Dtype:", X_train_scaled.dtype)
print("Min:", np.min(X_train_scaled))
print("Max:", np.max(X_train_scaled))
print("Finite:", np.isfinite(X_train_scaled).all())
print("y_train NaNs:", y_train.isna().sum())
print("y_train unique:", np.unique(y_train))
```

```python
lr = LogisticRegression(max_iter=2000, random_state=42)

lr.fit(X_train_scaled, y_train)

y_pred = lr.predict(X_test_scaled)
y_prob = lr.predict_proba(X_test_scaled)[:, 1]

auc = roc_auc_score(y_test, y_prob)

print("Logistic Regression")
print("Accuracy:", accuracy_score(y_test, y_pred))
print("ROC-AUC:", auc)
print("\nClassification Report:")
print(classification_report(y_test, y_pred))

fpr, tpr, _ = roc_curve(y_test, y_prob)

plt.figure(figsize=(7, 6))
plt.plot(fpr, tpr, label=f"Logistic Regression (AUC = {auc:.3f})")
plt.plot([0, 1], [0, 1], "--")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("Logistic Regression ROC Curve")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
svm = SVC(kernel="rbf", probability=True, random_state=42)

svm.fit(X_train_scaled, y_train)

y_pred = svm.predict(X_test_scaled)
y_prob = svm.predict_proba(X_test_scaled)[:, 1]

auc = roc_auc_score(y_test, y_prob)

print("SVM")
print("Accuracy:", accuracy_score(y_test, y_pred))
print("ROC-AUC:", auc)
print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# Confusion Matrix
cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(6, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Aspartic", "Serine"],
    yticklabels=["Aspartic", "Serine"]
)
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("SVM Confusion Matrix")
plt.show()

# ROC Curve
fpr, tpr, _ = roc_curve(y_test, y_prob)

plt.figure(figsize=(7, 6))
plt.plot(fpr, tpr, label=f"SVM (AUC = {auc:.3f})")
plt.plot([0, 1], [0, 1], "--")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("SVM ROC Curve")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```
